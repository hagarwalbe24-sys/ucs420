```python
x = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])
```

```python
x.shape
```

```python
x.ndim
```

```python
xt = x.T
xt
```

```python
a = xt @ x
a
```

```python
ai = np.linalg.inv(a)
ai
```

```python
b = ai @ xt @ y
b
```
