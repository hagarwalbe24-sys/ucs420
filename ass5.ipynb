```python
import numpy as np

# Array x
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
max_count = counts.max()
most_frequent = values[counts == max_count]

print("Array x:", x)
print("Most frequent value(s):", most_frequent)

for value in most_frequent:
    print("Indices of", value, ":", np.where(x == value)[0])


# Array y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
max_count = counts.max()
most_frequent = values[counts == max_count]

print("\nArray y:", y)
print("Most frequent value(s):", most_frequent)

for value in most_frequent:
    print("Indices of", value, ":", np.where(y == value)[0])
```

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# (a) First row, second column
print("1st row, 2nd column:", arr[0, 1])

# (b) Third row, first column
print("3rd row, 1st column:", arr[2, 0])
```

```python
import numpy as np

Harshil = np.linspace(10, 100, 25)

print("Array:", Harshil)
print("Dimensions:", Harshil.ndim)
print("Shape:", Harshil.shape)
print("Total elements:", Harshil.size)
print("Data type:", Harshil.dtype)
print("Total bytes consumed:", Harshil.nbytes)

# Reshape to a 5 x 5 array and transpose it
reshaped_Harshil = Harshil.reshape(5, 5)
transposed_Harshil = reshaped_Harshil.T

print("\nReshaped Array:\n", reshaped_Harshil)
print("\nTransposed Array:\n", transposed_Harshil)

# Can we use T directly on the original 1D array?
print("\nOriginal 1D array transpose:", Harshil.T)
```

```python
import numpy as np

ucs420_Harshil = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:\n", ucs420_Harshil)

print("\nMean:", np.mean(ucs420_Harshil))
print("Median:", np.median(ucs420_Harshil))
print("Maximum:", np.max(ucs420_Harshil))
print("Minimum:", np.min(ucs420_Harshil))
print("Unique Elements:", np.unique(ucs420_Harshil))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_Harshil = ucs420_Harshil.reshape(4, 3)

print("\nReshaped Array (4 x 3):\n",
      reshaped_ucs420_Harshil)

# Resize to 2 rows and 3 columns
resized_ucs420_Harshil = np.resize(ucs420_Harshil, (2, 3))

print("\nResized Array (2 x 3):\n",
      resized_ucs420_Harshil)
```
