```python
import random 
random.seed(1024170304)
hlo = []
for i in range (1,10):
    hlo.append(random.randint(1,100))
print(hlo)    


numbers = []

for i in range(100):
    numbers.append(random.randint(100, 900))

print(numbers)
count = 0 
for i in numbers : 
    if (i%2 == 1 ):
        print (i)
        count += 1

print (f" no of odds are {count}")
count = 0 
for i in numbers :
    if (i%2 == 0 ):
        print(i)
        count += 1

print (f"no of evens are {count}")
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True
prime = [x for x in numbers if is_prime(x)]
print (prime)

most = {}

for k in numbers:
    most[k] = most.get(k, 0) + 1

most_no = max(most, key=most.get)

print(f"The most frequent no is {most_no} and it occurs {most[most_no]} times")
```
