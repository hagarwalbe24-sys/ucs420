```python
import re
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

# Download required NLTK resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")

text = """
Artificial intelligence is changing the way people learn and work.
It helps computers solve problems and analyze large amounts of data.
Machine learning allows systems to improve through experience.
AI is used in healthcare, education, transportation, and business.
However, responsible AI must protect privacy and reduce bias.
I find artificial intelligence interesting because it combines logic and creativity.
"""

print("Original Text:\n", text)

clean_text = re.sub(r"[^\w\s]", "", text.lower())
print("\nLowercase Text Without Punctuation:\n", clean_text)

# 3. Tokenize into words and sentences
word_tokens = word_tokenize(clean_text)
sentence_tokens = sent_tokenize(text)

print("\nWord Tokens:\n", word_tokens)
print("\nSentence Tokens:\n", sentence_tokens)

# 4. Compare split() and word_tokenize()
split_tokens = clean_text.split()
nltk_tokens = word_tokenize(clean_text)

print("\nUsing split():\n", split_tokens)
print("\nUsing word_tokenize():\n", nltk_tokens)
print("\nNumber of split tokens:", len(split_tokens))
print("Number of NLTK tokens:", len(nltk_tokens))

# 5. Remove stopwords
stop_words = set(stopwords.words("english"))

filtered_words = [
    word for word in nltk_tokens
    if word.isalpha() and word not in stop_words
]

print("\nWords After Stopword Removal:\n", filtered_words)

# 6. Word frequency distribution
frequency = FreqDist(filtered_words)

print("\nWord Frequencies:")
for word, count in frequency.most_common():
    print(f"{word}: {count}")

# Plot word frequency
frequency.plot(10, title="Top 10 Words (Excluding Stopwords)")
plt.show()
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Download required resources
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("punkt")
nltk.download("punkt_tab")

# Same paragraph as Q1
text = """
Artificial intelligence is changing the way people learn and work.
It helps computers solve problems and analyze large amounts of data.
Machine learning allows systems to improve through experience.
AI is used in healthcare, education, transportation, and business.
However, responsible AI must protect privacy and reduce bias.
I find artificial intelligence interesting because it combines logic and creativity.
"""

# 1. Convert to lowercase and extract alphabetic words
words = re.findall(r"\b[a-zA-Z]+\b", text.lower())
print("Extracted Words:\n", words)

# 2. Remove stopwords
stop_words = set(stopwords.words("english"))
filtered_words = [word for word in words if word not in stop_words]

print("\nAfter Stopword Removal:\n", filtered_words)

# 3. Perform stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("\nStemmed Words:\n", stemmed_words)

# 4. Perform lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(word) for word in filtered_words
]

print("\nLemmatized Words:\n", lemmatized_words)

# 5. Compare results
print("\nComparison:")
for original, stem, lemma in zip(
    filtered_words, stemmed_words, lemmatized_words
):
    print(f"{original:15} -> Stem: {stem:12} Lemma: {lemma}")

print("\nConclusion:")
print("Stemming removes word endings using simple rules.")
print("Lemmatization returns dictionary forms of words.")
print("Lemmatization is preferable when meaningful words are important.")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# 1. Three short texts
texts = [
    "Artificial intelligence helps computers learn and solve problems.",
    "Machine learning uses data to make intelligent predictions.",
    "Artificial intelligence and machine learning are useful in healthcare."
]

for i, text in enumerate(texts, start=1):
    print(f"Text {i}: {text}")

# 2. Bag of Words representation
count_vectorizer = CountVectorizer(stop_words="english")
bow_matrix = count_vectorizer.fit_transform(texts)

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out(),
    index=["Text 1", "Text 2", "Text 3"]
)

print("\nBag of Words Matrix:")
print(bow_df)

# 3. TF-IDF representation
tfidf_vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out(),
    index=["Text 1", "Text 2", "Text 3"]
)

print("\nTF-IDF Matrix:")
print(tfidf_df.round(3))

# 4. Top 3 keywords from each text
feature_names = tfidf_vectorizer.get_feature_names_out()

print("\nTop 3 Keywords for Each Text:")
for i, row in enumerate(tfidf_matrix.toarray()):
    top_indices = row.argsort()[::-1][:3]

    keywords = [
        (feature_names[index], round(row[index], 3))
        for index in top_indices
        if row[index] > 0
    ]

    print(f"Text {i + 1}: {keywords}")
```

```python
import re
import nltk
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download("stopwords")

# 1. Two texts about different technologies
text1 = """
Artificial intelligence enables computers to learn from data.
It helps machines recognize patterns and make predictions.
AI is used in healthcare, education, and transportation.
It can automate tasks and support human decision making.
"""

text2 = """
Blockchain is a distributed digital ledger technology.
It stores transactions in linked blocks securely.
Blockchain is used in cryptocurrencies and supply chains.
It provides transparency and prevents unauthorized data changes.
"""

print("Text 1: Artificial Intelligence")
print(text1)

print("\nText 2: Blockchain")
print(text2)

# 2. Preprocess and tokenize
stop_words = set(stopwords.words("english"))

def preprocess(text):
    text = text.lower()
    words = re.findall(r"\b[a-z]+\b", text)
    return [word for word in words if word not in stop_words]

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("\nTokens 1:\n", tokens1)
print("\nTokens 2:\n", tokens2)

# 3a. Jaccard similarity
set1 = set(tokens1)
set2 = set(tokens2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard = len(intersection) / len(union) if union else 0

print("\nCommon Words:", intersection)
print("Jaccard Similarity:", round(jaccard, 4))

# 3b. Cosine similarity using TF-IDF
vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = vectorizer.fit_transform([text1, text2])

cosine = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)[0][0]

print("Cosine Similarity:", round(cosine, 4))

# 4. Compare the metrics
print("\nAnalysis:")
print("Jaccard measures the overlap between unique words.")
print("Cosine similarity compares the direction of TF-IDF vectors.")

if jaccard > cosine:
    print("Jaccard similarity is higher for these texts.")
elif cosine > jaccard:
    print("Cosine similarity is higher for these texts.")
else:
    print("Both similarity scores are equal.")

print(
    "For these technology descriptions, both metrics are useful. "
    "Jaccard is easy to interpret, while TF-IDF cosine similarity "
    "accounts for the relative importance of terms."
)
```

```python
import nltk
import matplotlib.pyplot as plt
from textblob import TextBlob
from wordcloud import WordCloud

nltk.download("punkt")
nltk.download("punkt_tab")

# 1. Product reviews
reviews = [
    "This phone has an excellent camera and amazing battery life.",
    "The product is fantastic and very easy to use.",
    "I love the quality and performance of this phone.",
    "The battery is terrible and the phone heats up quickly.",
    "Very disappointing product with poor performance.",
    "The delivery was slow and the product was damaged.",
    "The phone arrived yesterday.",
    "The box contains a phone and a charging cable."
]

# 2. Calculate polarity and subjectivity
positive_reviews = []

print("Sentiment Analysis Results:\n")

for i, review in enumerate(reviews, start=1):
    blob = TextBlob(review)

    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    # 3. Classify sentiment
    if polarity > 0:
        sentiment = "Positive"
        positive_reviews.append(review)
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print(f"Review {i}: {review}")
    print(f"Polarity: {polarity:.3f}")
    print(f"Subjectivity: {subjectivity:.3f}")
    print(f"Sentiment: {sentiment}\n")

# 4. Generate word cloud from positive reviews
if positive_reviews:
    positive_text = " ".join(positive_reviews)

    wordcloud = WordCloud(
        width=1000,
        height=500,
        background_color="white",
        stopwords={"the", "and", "is", "a", "to", "of"}
    ).generate(positive_text)

    plt.figure(figsize=(12, 6))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Word Cloud of Positive Reviews")
    plt.show()
else:
    print("No positive reviews available for the word cloud.")
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# 1. Training data (approximately 100 words)
text = """
Technology is changing the world every day.
Computers help people solve difficult problems.
Artificial intelligence allows machines to learn from data.
Machine learning improves predictions through experience.
Students use technology to study and explore new ideas.
Science and technology help doctors treat diseases.
Robots can perform repetitive tasks quickly and accurately.
The internet connects people across the world.
Digital tools make communication easier and faster.
Technology creates opportunities for learning and innovation.
People must use technology responsibly and ethically.
Artificial intelligence can support education and healthcare.
Computers analyze information and discover useful patterns.
New technologies can improve people's daily lives.
Learning technology helps students prepare for the future.
"""

# 2. Tokenize text using Keras Tokenizer
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Total unique words:", total_words)
print("\nWord Index:")
print(tokenizer.word_index)

# 3. Create input sequences
input_sequences = []

for line in text.strip().split("\n"):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(2, len(token_list) + 1):
        input_sequences.append(token_list[:i])

# Pad sequences
max_sequence_len = max(len(seq) for seq in input_sequences)

input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding="pre"
)

# Inputs: all tokens except the final token
X = input_sequences[:, :-1]

# Target: final token
y = input_sequences[:, -1]

# 4. Build a simple LSTM model
model = Sequential([
    Embedding(
        input_dim=total_words,
        output_dim=32,
        input_length=max_sequence_len - 1
    ),
    LSTM(64),
    Dense(total_words, activation="softmax")
])

model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.summary()

# 5. Train the model
history = model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)

# 6. Generate new text from a seed phrase
def generate_text(seed_text, next_words=10):
    generated_text = seed_text.lower()

    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences(
            [generated_text]
        )[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding="pre"
        )

        predictions = model.predict(token_list, verbose=0)[0]

        predicted_id = int(np.argmax(predictions))

        if predicted_id == 0:
            break

        output_word = tokenizer.index_word.get(predicted_id)

        if output_word is None:
            break

        generated_text += " " + output_word

    return generated_text

# Generate 2-3 new lines
print("\nGenerated Text:")
print(generate_text("technology", 10))
print(generate_text("artificial intelligence", 10))
print(generate_text("students", 10))
```
